```python
class ImageDataset(Dataset):
    def __init__(self, data_dir):
        self.data_dir = data_dir

    def __len__(self):
        return 1

    def __getitem__(self, idx):
      device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

      vector_field1_path = f'{self.data_dir}/sin/sin_vector_field.nhdr'
      mask1_path = f'{self.data_dir}/sin/sin_filt_mask.nhdr'
      vector_field1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field1_path))).permute(2,0,1).to(device).float()*1000.0
      mask1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask1_path))).permute(1,0)
      # Convert to numpy BEFORE passing to laplace
      boundary_mask1 = torch.where(torch.from_numpy(filters.laplace(mask1.cpu().numpy()))>0,1,0)
      mask1 = (mask1-boundary_mask1).float().to(device)

      vector_field2_path = f'{self.data_dir}/cos/cos_vector_field.nhdr'
      mask2_path = f'{self.data_dir}/cos/cos_filt_mask.nhdr'
      vector_field2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field2_path))).permute(2,0,1).to(device).float()*1000.0
      mask2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask2_path))).permute(1,0)
      # Convert to numpy BEFORE passing to laplace
      boundary_mask2 = torch.where(torch.from_numpy(filters.laplace(mask2.cpu().numpy()))>0,1,0)
      mask2 = (mask2-boundary_mask2).float().to(device)

      sample = {  'vector_field'  : torch.cat((vector_field1,vector_field2),0),
                  'mask1'          : mask1.unsqueeze(0),
                  'mask2'          : mask2.unsqueeze(0)}
      return sample

    # def __getitem__(self, idx):
    #     device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    #     vector_field1_path = f'{self.data_dir}/sin/sin_vector_field.nhdr'
    #     mask1_path = f'{self.data_dir}/sin/sin_filt_mask.nhdr'
    #     vector_field1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field1_path))).permute(2,0,1).to(device).float()*1000.0
    #     mask1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask1_path))).permute(1,0)
    #     boundary_mask1 = torch.where(torch.from_numpy(filters.laplace(mask1))>0,1,0)
    #     mask1 = (mask1-boundary_mask1).float().to(device)

    #     vector_field2_path = f'{self.data_dir}/cos/cos_vector_field.nhdr'
    #     mask2_path = f'{self.data_dir}/cos/cos_filt_mask.nhdr'
    #     vector_field2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field2_path))).permute(2,0,1).to(device).float()*1000.0
    #     mask2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask2_path))).permute(1,0)
    #     boundary_mask2 = torch.where(torch.from_numpy(filters.laplace(mask2))>0,1,0)
    #     mask2 = (mask2-boundary_mask2).float().to(device)

    #     sample = {  'vector_field'  : torch.cat((vector_field1,vector_field2),0),
    #                 'mask1'          : mask1.unsqueeze(0),
    #                 'mask2'          : mask2.unsqueeze(0)}
    #     return sample

def matrix_exp_2d(A):
    """
    Construct positive definite matrix from symmetric matrix field A
    Args:
        A, torch.Tensor
    Returns:
        psd, torch.Tensor
    """
    I = torch.zeros_like(A, device='cuda')
    I[...,0,0] = 1
    I[...,1,1] = 1

    s = ((A[...,0,0]+A[...,1,1])/2.).unsqueeze(-1).unsqueeze(-1)
    q = torch.sqrt(-torch.det(A-torch.mul(s, I))).unsqueeze(-1).unsqueeze(-1)

    psd = torch.exp(s)*(torch.mul((torch.cosh(q)-s*torch.sinh(q)/q),I)+torch.sinh(q)/q*A)
    return psd

def pde(u, vector_lin, mask, differential_accuracy=2):
    s = tensors.lin2mat(u)
    metric_mat = matrix_exp_2d(s)
    nabla_vv = riemann.covariant_derivative_2d(vector_lin, metric_mat, mask, differential_accuracy=differential_accuracy)

    return nabla_vv
```

### [Matrix exponential for 2D](https://en.wikipedia.org/wiki/Matrix_exponential)
${\displaystyle e^{tA}=e^{st}\left(\left(\cosh(qt)-s{\frac {\sinh(qt)}{q}}\right)~I~+{\frac {\sinh(qt)}{q}}A\right)~.}$ where $t=1$, $s=tr({A}/2), q=-\sqrt{-\det(A-sI)}$

```python
epoch_loss_list = []
epoch_num = 1001
start_epoch_num = 1001
learning_rate = 1e-2

model = DenseED(in_channels=4, out_channels=3,
                imsize=100,
                blocks=[6, 8, 6],
                growth_rate=16,
                init_features=48,
                drop_rate=0,
                out_activation=None,
                upsample='nearest')
model.train()
if mode=='gpu':
    model.cuda()
criterion = torch.nn.MSELoss()
optimizer_id = torch.optim.Adadelta(model.parameters(), lr=learning_rate)

dataset_id = ImageDataset(input_dir)
dataloader_id = DataLoader(dataset_id, batch_size=1, shuffle=False, num_workers=0)

if resume:
    checkpoint = torch.load(f'{output_dir}/epoch_{start_epoch_num-1}_checkpoint.pth.tar')
    model.load_state_dict(checkpoint['model_state_dict'])

    with open(f'{output_dir}/loss.txt', 'a') as f:
        f.write(f'From {start_epoch_num} to {epoch_num+start_epoch_num}\n')
        f.write(f'MSE; Adadelta, lr={learning_rate};\n')
else:
    start_epoch_num = 0

    with open(f'{output_dir}/loss.txt', 'w+') as f:
        f.write(f'From {start_epoch_num} to {epoch_num+start_epoch_num}\n')
        f.write(f'MSE; Adadelta: lr={learning_rate};\n')

print(f'Starting from iteration {start_epoch_num} to iteration {epoch_num+start_epoch_num}')
```

```python
for epoch in tqdm(range(start_epoch_num, start_epoch_num+epoch_num)):
    epoch_loss_id = 0
    epoch_loss_bd = 0

    for i, batched_id_sample in enumerate(dataloader_id):
        '''innerdomain backpropagate'''
        input_id = batched_id_sample['vector_field'].to(device)
        input_id.requires_grad = True
        # u_pred_id: [batch_size, *data_shape, feature_num] = [1, 3, ...]
        u_pred_id = model(input_id)
        # f_pred_id: [batch_size, feature_num, *data_shape] = [1, 2, ...]
        mask1 = batched_id_sample['mask1'].squeeze()
        mask2 = batched_id_sample['mask2'].squeeze()
        pde_loss1 = pde(u_pred_id.squeeze(), input_id[0,:2].squeeze(), mask1, differential_accuracy=2)
        pde_loss2 = pde(u_pred_id.squeeze(), input_id[0,2:].squeeze(), mask2, differential_accuracy=2)
        f_pred_1 = torch.einsum('...ij,...ij->...ij', pde_loss1, mask1.unsqueeze(0).expand(2,-1,-1))
        f_pred_2 = torch.einsum('...ij,...ij->...ij', pde_loss2, mask2.unsqueeze(0).expand(2,-1,-1))
        f_pred_id = torch.cat((f_pred_1,f_pred_2),0)
        # f_true_id: [batch_size, feature_num, *data_shape] = [1, 2, ...]
        f_true_id = torch.zeros_like(f_pred_id)

        optimizer_id.zero_grad()
        loss_id = criterion(f_pred_id, f_true_id)
        loss_id.backward()
        epoch_loss_id += loss_id.item()
        optimizer_id.step()

    with open(f'{output_dir}/loss.txt', 'a') as f:
        f.write(f'{epoch_loss_id}\n')

    print(f'epoch {epoch} innerdomain loss: {epoch_loss_id}, norm: {torch.norm(f_pred_id,2)**2}')
    epoch_loss_list.append(epoch_loss_id)
    if epoch%10==0:
        if save_model:
            torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_id_state_dict': optimizer_id.state_dict(),
            'loss_id': epoch_loss_id
            }, f'{output_dir}/model.pth.tar')
```

```python

# Inference to get learned metric on a given vector field
# model must be in_channels=4, bottleneck=True, growth_rate=32

def infer_metric(vf_4ch):
    with torch.no_grad():
        u_pred = model(vf_4ch.unsqueeze(0).float())   # (1,3,H,W)
        s_pred = tensors.lin2mat(u_pred.squeeze())    # (H,W,2,2)
        metric_mat = torch.linalg.matrix_exp(s_pred)
    return metric_mat

# Helpers
def dilate_mask(mask, it=2):
    m = mask.numpy() if torch.is_tensor(mask) else mask
    m = binary_dilation(m > 0.5, iterations=it).astype(float)
    return torch.from_numpy(m) if torch.is_tensor(mask) else m

def find_valid_start_points(mask, vf, num_points=5, v_thresh=1e-3):
    m = mask.numpy()
    v = vf.numpy()
    coords = np.where(m > 0.5)
    if len(coords[0]) == 0: return []
    idxs = np.linspace(0, len(coords[0])-1, num_points*4).astype(int)
    pts = []
    for i in idxs:
        pt = np.array([coords[0][i], coords[1][i]])
        if np.linalg.norm(v[:, pt[0], pt[1]]) > v_thresh:
            pts.append(pt)
        if len(pts) >= num_points: break
    return pts
```

##1) Experiment A — Curvature Stress Test (frequency scaling)
Generate braids with increasing curvature (freq multipliers) using paper formulas, keep fibers thick, run inference, and measure PDE loss + anisotropy.

```python
freq_levels = [1.0, 2.0, 3.0, 4.0]

def generate_braid(freq=1.0, img_size=100, fiber_width=7):
    base_freq = np.pi / 4 * freq
    x1 = np.arange(img_size)
    # Paper curves
    y1 = 20 * np.cos(base_freq * (x1 - 60)) + 50
    y2 = 20 * np.sin(base_freq * x1) + 50
    dy1 = -20 * base_freq * np.sin(base_freq * (x1 - 60))
    dy2 =  20 * base_freq * np.cos(base_freq * x1)

    vf = np.zeros((4, img_size, img_size))
    m1 = np.zeros((img_size, img_size))
    m2 = np.zeros((img_size, img_size))

    for i, x in enumerate(x1):
        for (y, dy, c0, c1) in [(y1[i], dy1[i], 0, 1), (y2[i], dy2[i], 2, 3)]:
            y = int(np.clip(y, 0, img_size-1))
            t = np.array([1.0, dy]); t = t / (np.linalg.norm(t)+1e-8)
            for d in range(-fiber_width, fiber_width+1):
                yy = y + d
                if 0 <= yy < img_size:
                    w = np.exp(-(d**2)/(fiber_width**2))
                    if c0 == 0: m1[x, yy] = max(m1[x, yy], w)
                    else:       m2[x, yy] = max(m2[x, yy], w)
                    vf[c0, x, yy] += t[0]*w
                    vf[c1, x, yy] += t[1]*w

    for c in [0,2]:
        norm = np.sqrt(vf[c]**2 + vf[c+1]**2)
        norm = np.where(norm>1e-6, norm, 1)
        vf[c] /= norm; vf[c+1] /= norm

    return {
        'vector_field': torch.from_numpy(vf).float(),
        'mask1': torch.from_numpy((m1>0.3).astype(float)),
        'mask2': torch.from_numpy((m2>0.3).astype(float)),
        'combined_mask': torch.from_numpy(np.clip((m1>0.3)+(m2>0.3),0,1).astype(float))
    }

datasets = {f: generate_braid(f) for f in freq_levels}
device = torch.device('cuda')

# Compute metrics: PDE loss proxy via mean |∇_V V|
def covariant_error_gpu(metric_mat, vf, mask):
    # all tensors on the same device
    nabla = riemann.covariant_derivative_2d(vf, metric_mat, mask)
    ce = torch.sqrt(nabla[0]**2 + nabla[1]**2)
    return ce[mask > 0.5].mean().item()

results = {}
for f in freq_levels:
    vf_all = datasets[f]['vector_field'].to(device)  # (4,H,W) on GPU
    mask1  = datasets[f]['mask1'].to(device)
    metric = infer_metric(vf_all)                    # stays on GPU

    ce = covariant_error_gpu(metric, vf_all[:2], mask1)

    eigs    = torch.linalg.eigvalsh(metric)
    lam_min = eigs[...,0][mask1 > 0.5]
    lam_max = eigs[...,1][mask1 > 0.5]
    aniso   = (lam_max / (lam_min + 1e-8)).mean().item()

    results[f] = {'cov_error': ce, 'anisotropy': aniso}
    print(f"Freq {f}x: cov_error={ce:.3e}, anisotropy={aniso:.2f}")
```

```python
freqs = np.array(sorted(results.keys()))
covs  = np.array([results[f]['cov_error']  for f in freqs])
anis  = np.array([results[f]['anisotropy'] for f in freqs])

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(freqs, covs, 'o-', lw=2, ms=8, color='#E63946')
axes[0].set_xlabel('Frequency multiplier')
axes[0].set_ylabel('Covariant error ||∇_V V||')
axes[0].set_title('Curvature Stress: Covariant Error')
axes[0].grid(True, alpha=0.3)

axes[1].plot(freqs, anis, 'o-', lw=2, ms=8, color='#2A9D8F')
axes[1].set_xlabel('Frequency multiplier')
axes[1].set_ylabel('Anisotropy λ_max/λ_min')
axes[1].set_title('Curvature Stress: Anisotropy')
axes[1].grid(True, alpha=0.3)

plt.suptitle('Curvature Stress Test', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('../plots/curvature_stress_trends.png', dpi=150)
plt.show()
```

## 2) Experiment B — Fiber Thickness Robustness
Keep curvature fixed (1.0×), vary fiber thickness, measure geodesic survival (how many steps stay inside mask)

```python

def compute_geodesic_2d(metric_lin, vector_lin, mask, start_point, initial_velocity=None,
                        delta_t=0.15, max_iter=2000):
    """Compute geodesic using the geodesic equation γ̈ + Γ(γ̇, γ̇) = 0"""

    # Convert to numpy if needed
    if torch.is_tensor(metric_lin):
        metric_lin = metric_lin.cpu().detach().numpy()
    if torch.is_tensor(vector_lin):
        vector_lin = vector_lin.cpu().detach().numpy()
    if torch.is_tensor(mask):
        mask = mask.cpu().numpy()

    H, W = mask.shape
    metric_mat = tensors.lin2mat(metric_lin)

    # Compute Christoffel symbols
    Gamma1, Gamma2 = riemann.get_christoffel_symbol_2d(metric_mat, mask)

    path_x, path_y = [], []
    gamma = np.array(start_point, dtype=np.float64)

    # Initial velocity from vector field
    if initial_velocity is None:
        x_int, y_int = int(gamma[0]), int(gamma[1])
        if 0 <= x_int < H and 0 <= y_int < W:
            v = np.array([vector_lin[0, x_int, y_int], vector_lin[1, x_int, y_int]])
            v_norm = np.linalg.norm(v)
            gamma_dot = v / v_norm if v_norm > 1e-6 else np.array([1.0, 0.0])
        else:
            gamma_dot = np.array([1.0, 0.0])
    else:
        gamma_dot = np.array(initial_velocity, dtype=np.float64)

    path_x.append(gamma[0])
    path_y.append(gamma[1])

    for i in range(max_iter):
        x, y = gamma
        xi, yi = int(np.clip(x, 0, H-1)), int(np.clip(y, 0, W-1))

        if mask[xi, yi] < 0.5 or x < 1 or x > H-2 or y < 1 or y > W-2:
            break

        # Bilinear interpolation of Christoffel symbols
        x0, y0 = int(x), int(y)
        x1, y1 = min(x0+1, H-1), min(y0+1, W-1)
        fx, fy = x - x0, y - y0

        def interp(field):
            return ((1-fx)*(1-fy)*field[x0, y0] + fx*(1-fy)*field[x1, y0] +
                    (1-fx)*fy*field[x0, y1] + fx*fy*field[x1, y1])

        G1 = np.array([[interp(Gamma1[..., 0, 0]), interp(Gamma1[..., 0, 1])],
                       [interp(Gamma1[..., 1, 0]), interp(Gamma1[..., 1, 1])]])
        G2 = np.array([[interp(Gamma2[..., 0, 0]), interp(Gamma2[..., 0, 1])],
                       [interp(Gamma2[..., 1, 0]), interp(Gamma2[..., 1, 1])]])

        # Geodesic equation
        gamma_ddot = np.zeros(2)
        gamma_ddot[0] = -np.einsum('ij,i,j->', G1, gamma_dot, gamma_dot)
        gamma_ddot[1] = -np.einsum('ij,i,j->', G2, gamma_dot, gamma_dot)

        gamma_dot = gamma_dot + delta_t * gamma_ddot
        gamma = gamma + delta_t * gamma_dot

        path_x.append(gamma[0])
        path_y.append(gamma[1])

    return np.array(path_x), np.array(path_y)
```

```python
from scipy.ndimage import binary_dilation, binary_erosion

thickness_levels = [0.5, 0.75, 1.0, 1.5, 2.0]

def adjust_thickness(mask, factor):
    m = mask.detach().cpu().numpy()
    if factor > 1:
        m = binary_dilation(m > 0.5, iterations=int((factor-1)*3)).astype(float)
    elif factor < 1:
        m = binary_erosion(m > 0.5, iterations=int((1-factor)*3)).astype(float)
    return torch.from_numpy(m)

def geodesic_steps_inside(metric_lin, vf_2ch, mask, start, delta_t=0.05, max_iter=800):
    gx, gy = compute_geodesic_2d(metric_lin, vf_2ch, mask, start, delta_t=delta_t, max_iter=max_iter)
    m = mask.detach().cpu().numpy()
    inside = 0
    for x, y in zip(gx, gy):
        xi, yi = int(np.clip(x, 0, 99)), int(np.clip(y, 0, 99))
        if m[xi, yi] > 0.5:
            inside += 1
        else:
            break
    return inside

print("\nTHICKNESS ROBUSTNESS")
for t in thickness_levels:
    d = generate_braid(freq=1.0)                # 4ch vector field, masks on CPU
    mask_t = adjust_thickness(d['mask1'], t)    # CPU
    vf_all = d['vector_field'].to(device)       # 4ch -> GPU for inference

    # infer metric on GPU, then move to CPU for geodesic integration
    metric = infer_metric(vf_all).cpu()
    metric_lin = tensors.mat2lin(metric.detach().numpy())

    vf_f1 = d['vector_field'][:2].cpu()         # fiber 1 on CPU for geodesics
    seeds = find_valid_start_points(mask_t, vf_f1, num_points=3, v_thresh=1e-3)
    steps = []
    for s in seeds:
        steps.append(geodesic_steps_inside(metric_lin, vf_f1, mask_t, s))
    print(f"Thickness {t}x: avg steps inside = {np.mean(steps):.1f}")
```

## 3) Experiment C — Geodesic Parallelism (Stability)
Measure divergence (Hausdorff) between geodesics from nearby seeds on each curvature level.

```python
parallelism_results = {}

def hausdorff(pts1, pts2):
    d1 = np.max([np.min(np.sqrt(np.sum((pts2 - p)**2, axis=1))) for p in pts1])
    d2 = np.max([np.min(np.sqrt(np.sum((pts1 - p)**2, axis=1))) for p in pts2])
    return max(d1, d2)

print("\nGEODESIC PARALLELISM")
for f in freq_levels:
    d = datasets[f]
    # infer metric on GPU, then move to CPU for geodesics
    metric = infer_metric(d['vector_field'].to(device)).cpu().numpy()
    metric_lin = tensors.mat2lin(metric)

    mask = dilate_mask(d['mask1'], it=2).cpu()
    vf_f1 = d['vector_field'][:2].cpu()
    seeds = find_valid_start_points(mask, vf_f1, num_points=5, v_thresh=1e-3)

    geos = []
    for s in seeds:
        gx, gy = compute_geodesic_2d(metric_lin, vf_f1.numpy(), mask.numpy(), s, delta_t=0.05, max_iter=800)
        if len(gx) > 20:
            geos.append(np.column_stack([gx, gy]))

    if len(geos) < 2:
        print(f"Freq {f}x: not enough geodesics")
        parallelism_results[f] = {'divergence': np.nan}
        continue

    dists = []
    for i in range(len(geos)):
        for j in range(i+1, len(geos)):
            dists.append(hausdorff(geos[i], geos[j]))
    div = np.mean(dists)
    parallelism_results[f] = {'divergence': div}
    print(f"Freq {f}x: divergence (Hausdorff) = {div:.2f} px")
```

```python
import numpy as np, torch, matplotlib.pyplot as plt
from util import riemann
import util.tensors as tensors

device = torch.device('cuda')

def compute_geodesic_curvature_gpu(metric_lin, vector_lin, mask):
    # All inputs must already be on GPU
    metric_mat = tensors.lin2mat(metric_lin)            # GPU
    nabla = riemann.covariant_derivative_2d(vector_lin, metric_mat, mask)
    nabla_norm = torch.sqrt(nabla[0]**2 + nabla[1]**2)
    v_norm_sq  = vector_lin[0]**2 + vector_lin[1]**2 + 1e-8
    return nabla_norm / v_norm_sq   # GPU tensor

fig, axes = plt.subplots(2, len(freq_levels), figsize=(16, 6))
for idx, f in enumerate(freq_levels):
    # Inference on GPU
    vf_all = datasets[f]['vector_field'].to(device)   # (4,H,W) GPU
    mask1  = datasets[f]['mask1'].to(device)          # GPU
    metric = infer_metric(vf_all)                     # GPU
    metric_lin = tensors.mat2lin(metric)              # GPU

    # κ_geo on GPU
    kappa_geo = compute_geodesic_curvature_gpu(metric_lin, vf_all[:2], mask1)

    # Move to CPU ONLY for plotting
    kappa_geo_np = kappa_geo.detach().cpu().numpy()
    mask_np      = mask1.detach().cpu().numpy()

    # True curvature (numpy) for the paper curve
    x1 = np.arange(100)
    base_freq = np.pi/4 * f
    y1  = 20*np.cos(base_freq*(x1-60))+50
    dy  = -20*base_freq*np.sin(base_freq*(x1-60))
    dy2 = -20*(base_freq**2)*np.cos(base_freq*(x1-60))
    k_true = np.abs(dy2)/(1+dy**2)**1.5

    # Heatmap
    ax = axes[0, idx]
    ax.imshow((kappa_geo_np * mask_np).T, origin='lower', cmap='hot',
              vmin=0, vmax=np.nanmax(kappa_geo_np))
    ax.set_title(f'Freq {f}x\nκ_geo'); ax.axis('off')

    # Line plot along the curve
    k_geo_samples = []
    for i, x in enumerate(x1):
        y = int(np.clip(y1[i], 0, 99))
        k_geo_samples.append(kappa_geo_np[x, y] if mask_np[x, y] > 0.5 else np.nan)

    ax = axes[1, idx]
    ax.plot(x1, k_true, 'b-', label='κ_true')
    ax.plot(x1, k_geo_samples, 'r--', label='κ_geo', alpha=0.7)
    ax.set_ylim(0, max(k_true)*1.6); ax.set_xlabel('x'); ax.set_ylabel('curvature')
    if idx == 0: ax.legend()

plt.suptitle('Curvature Fidelity: κ_geo vs κ_true', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('../plots/curvature_fidelity.png', dpi=150)
plt.show()
```

```python
import numpy as np
import torch
import matplotlib.pyplot as plt

def generate_paper_braid(freq_multiplier=1.0, img_size=100):
    """
    Generate braids using the EXACT paper formulas, with optional frequency scaling.

    Paper formulas (freq_multiplier=1.0):
        Curve 1: x₂ = 20·cos(π/4·(x₁ - 60)) + 50
        Curve 2: x₂ = 20·sin(π/4·x₁) + 50

    Stress test: increase freq_multiplier to 2, 3, 4 for sharper curves
    """

    # Base frequency from paper
    base_freq = np.pi / 4
    freq = base_freq * freq_multiplier

    # Spatial coordinates
    x1 = np.arange(img_size)

    # Paper's exact curves
    curve1_x2 = 20 * np.cos(freq * (x1 - 60)) + 50  # Paper curve 1
    curve2_x2 = 20 * np.sin(freq * x1) + 50          # Paper curve 2

    # Derivatives (tangent directions)
    curve1_dx2 = -20 * freq * np.sin(freq * (x1 - 60))  # dy/dx for curve 1
    curve2_dx2 = 20 * freq * np.cos(freq * x1)           # dy/dx for curve 2

    # Tangent vectors (dx1=1 since we parameterize by x1)
    # For curve 1: tangent = (1, curve1_dx2)
    # For curve 2: tangent = (1, curve2_dx2)

    # Create image-sized arrays
    vf = np.zeros((4, img_size, img_size))
    mask1 = np.zeros((img_size, img_size))
    mask2 = np.zeros((img_size, img_size))

    # Fiber width (paper uses 9 pixels horizontal)
    fiber_width = 7  # radius in pixels

    # Draw curve 1 bundle
    for i, x in enumerate(x1):
        y_center = int(np.clip(curve1_x2[i], 0, img_size-1))
        tangent = np.array([1.0, curve1_dx2[i]])
        tangent = tangent / np.linalg.norm(tangent)  # normalize

        for dy in range(-fiber_width, fiber_width+1):
            y = y_center + dy
            if 0 <= y < img_size:
                weight = np.exp(-(dy**2) / (fiber_width**2))
                mask1[x, y] = max(mask1[x, y], weight)
                vf[0, x, y] += tangent[0] * weight
                vf[1, x, y] += tangent[1] * weight

    # Draw curve 2 bundle
    for i, x in enumerate(x1):
        y_center = int(np.clip(curve2_x2[i], 0, img_size-1))
        tangent = np.array([1.0, curve2_dx2[i]])
        tangent = tangent / np.linalg.norm(tangent)

        for dy in range(-fiber_width, fiber_width+1):
            y = y_center + dy
            if 0 <= y < img_size:
                weight = np.exp(-(dy**2) / (fiber_width**2))
                mask2[x, y] = max(mask2[x, y], weight)
                vf[2, x, y] += tangent[0] * weight
                vf[3, x, y] += tangent[1] * weight

    # Normalize vector fields
    for c in [0, 2]:
        norm = np.sqrt(vf[c]**2 + vf[c+1]**2)
        norm = np.where(norm > 1e-6, norm, 1)
        vf[c] /= norm
        vf[c+1] /= norm

    # Binary masks
    mask1_bin = (mask1 > 0.3).astype(float)
    mask2_bin = (mask2 > 0.3).astype(float)

    # Compute crossing angle at center
    center_idx = img_size // 2
    slope1 = curve1_dx2[center_idx]
    slope2 = curve2_dx2[center_idx]
    angle_rad = np.arctan(np.abs((slope1 - slope2) / (1 + slope1 * slope2 + 1e-6)))
    crossing_angle = np.degrees(angle_rad)

    # # At the end, before returning, add small noise to avoid exact zeros:
    # vf = np.where(np.abs(vf) < 1e-6, 1e-6, vf)  # Replace zeros with tiny values

    # return {
    #     'vector_field': torch.from_numpy(vf).float(),
    #     # ... rest ...
    # }

    return {
        'vector_field': torch.from_numpy(vf).float(),
        'mask1': torch.from_numpy(mask1_bin).float(),
        'mask2': torch.from_numpy(mask2_bin).float(),
        'combined_mask': torch.from_numpy(np.clip(mask1_bin + mask2_bin, 0, 1)).float(),
        'freq_multiplier': freq_multiplier,
        'approx_crossing_angle': crossing_angle,
        'curve1_y': curve1_x2,
        'curve2_y': curve2_x2
    }

# =====================================================
# EXPERIMENT: Paper reproduction + Stress test
# =====================================================

freq_levels = [1.0, 2.0, 3.0, 4.0]  # 1.0 = paper's exact curves
datasets = {}

print("Generating braids with paper formulas + frequency scaling:")
print("-" * 50)
for freq in freq_levels:
    datasets[freq] = generate_paper_braid(freq_multiplier=freq)
    angle = datasets[freq]['approx_crossing_angle']
    print(f"  Freq {freq}x: Approx crossing angle ≈ {angle:.1f}°")
```

```python
from util import riemann

def pde_safe(u, vector_lin, mask, differential_accuracy=2):
    """
    PDE loss with epsilon fix to prevent division by zero.
    """
    s = tensors.lin2mat(u)
    metric_mat = matrix_exp_2d(s)

    nabla_vv = riemann.covariant_derivative_2d(vector_lin, metric_mat, mask, differential_accuracy)

    # FIXED: Add epsilon to prevent NaN
    v_squared = vector_lin[0]**2 + vector_lin[1]**2 + 1e-8  # <-- THE FIX
    sigma = (vector_lin[0]*nabla_vv[0] + vector_lin[1]*nabla_vv[1]) / v_squared

    return torch.stack((nabla_vv[0] - sigma*vector_lin[0],
                        nabla_vv[1] - sigma*vector_lin[1]), 0)

# Replace the imported pde with our safe version
pde = pde_safe
print("✓ Using patched pde() function with NaN fix")
```

```python

# ============================================================
# STEP 2: Training function
# ============================================================

def train_and_evaluate(data, num_epochs=500, device='cuda'):
    """Train CEDNN and return performance metrics."""

    model = DenseED(in_channels=4, out_channels=3, imsize=100,
                    blocks=[6, 8, 6], growth_rate=16, init_features=48,
                    drop_rate=0, out_activation=None, upsample='nearest',
                    bottleneck=True)
    model = model.to(device)
    model.train()

    # optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    # Replace Adam with Adadelta (what the original braid notebook uses)
    optimizer = torch.optim.Adadelta(model.parameters(), lr=1e-2)
    criterion = torch.nn.MSELoss()

    vf = data['vector_field'].unsqueeze(0).to(device)
    mask1 = data['mask1'].to(device)
    mask2 = data['mask2'].to(device)

    loss_history = []

    for epoch in tqdm(range(num_epochs), desc=f"Training freq={data['freq_multiplier']}x"):
        vf.requires_grad = True
        u_pred = model(vf)

        pde_loss1 = pde(u_pred.squeeze(), vf[0,:2].squeeze(), mask1, differential_accuracy=2)
        pde_loss2 = pde(u_pred.squeeze(), vf[0,2:].squeeze(), mask2, differential_accuracy=2)

        f_pred_1 = pde_loss1 * mask1.unsqueeze(0).expand(2,-1,-1)
        f_pred_2 = pde_loss2 * mask2.unsqueeze(0).expand(2,-1,-1)
        f_pred = torch.cat((f_pred_1, f_pred_2), 0)

        f_true = torch.zeros_like(f_pred)

        optimizer.zero_grad()
        loss = criterion(f_pred, f_true)
        loss.backward()
        optimizer.step()

        loss_history.append(loss.item())

    # Get final metric
    model.eval()
    with torch.no_grad():
        u_pred = model(vf)
        s_pred = tensors.lin2mat(u_pred.squeeze())
        metric_mat = matrix_exp_2d(s_pred)

    return model, metric_mat.cpu(), loss_history


def analyze_metric_quality(metric_mat, mask):
    """Compute stability metrics for the learned metric."""
    metric_np = metric_mat.numpy()
    mask_np = mask.numpy()

    eigenvalues = np.zeros((metric_np.shape[0], metric_np.shape[1], 2))

    for i in range(metric_np.shape[0]):
        for j in range(metric_np.shape[1]):
            if mask_np[i, j] > 0.5:
                g = metric_np[i, j]
                try:
                    eigs = np.linalg.eigvalsh(g)
                    eigenvalues[i, j] = np.sort(eigs)
                except:
                    eigenvalues[i, j] = [1, 1]

    lambda_min = eigenvalues[..., 0]
    lambda_max = eigenvalues[..., 1]
    lambda_min = np.where(lambda_min > 1e-6, lambda_min, 1e-6)

    anisotropy = lambda_max / lambda_min
    condition_number = lambda_max / lambda_min

    # Stats in masked region
    valid = mask_np > 0.5

    return {
        'mean_anisotropy': np.mean(anisotropy[valid]),
        'std_anisotropy': np.std(anisotropy[valid]),
        'mean_lambda_min': np.mean(lambda_min[valid]),
        'mean_lambda_max': np.mean(lambda_max[valid]),
        'mean_condition': np.mean(condition_number[valid]),
        'max_condition': np.max(condition_number[valid]),
        'anisotropy_map': anisotropy,
        'eigenvalues': eigenvalues
    }
```

#EVALUATE GEODESCICS

```python
# ============================================================
# GEODESIC VERIFICATION: Do learned metrics recover ground truth?
# ============================================================
import warnings
warnings.filterwarnings('ignore', category=SyntaxWarning)
import algo.geodesic as geo
import algo.euler as euler

def evaluate_geodesic_alignment(metric_mat, vector_field, mask, start_point,
                                  delta_t=5e-2, num_iters=30000):
    """
    Compare geodesic (from learned metric) vs integral curve (ground truth).
    Returns mean-min error.
    """
    # Convert metric to tensor (inverse)
    metric_np = metric_mat.numpy()
    tensor_mat = np.linalg.inv(metric_np)
    tensor_lin = tensors.mat2lin(tensor_mat)

    vf_np = vector_field.numpy()
    mask_np = mask.numpy()

    # Initial velocity from vector field
    init_vel = vf_np[:, int(start_point[0]), int(start_point[1])]

    # Compute geodesic
    try:
        geo_x, geo_y = geo.geodesicpath('f', tensor_lin, vf_np, mask_np,
                                         start_point, init_vel, delta_t,
                                         iter_num=num_iters, both_directions=True)
    except:
        return float('inf'), [], []

    # Compute integral curve (ground truth)
    try:
        int_x, int_y = euler.eulerpath_vectbase_2d_w_dv(vf_np, mask_np, start_point,
                                                         delta_t, iter_num=num_iters,
                                                         both_directions=True)
    except:
        return float('inf'), [], []

    # Mean-min error
    if len(geo_x) == 0 or len(int_x) == 0:
        return float('inf'), geo_x, int_x

    geo_pts = np.column_stack([geo_x, geo_y])
    int_pts = np.column_stack([int_x, int_y])

    min_dists = []
    for gpt in geo_pts:
        dists = np.sqrt(np.sum((int_pts - gpt)**2, axis=1))
        min_dists.append(np.min(dists))

    return np.mean(min_dists), (geo_x, geo_y), (int_x, int_y)
# ============================================================
# GEODESIC VERIFICATION - WITH VALID STARTING POINTS
# ============================================================

def find_valid_start_points(mask, num_points=3):
    """Find points that are actually inside the mask (on the fibers)."""
    mask_np = mask.numpy()
    valid_coords = np.where(mask_np > 0.5)

    if len(valid_coords[0]) == 0:
        return []

    # Sample evenly spaced points along the fiber
    indices = np.linspace(0, len(valid_coords[0])-1, num_points).astype(int)
    points = [np.array([valid_coords[0][i], valid_coords[1][i]]) for i in indices]

    return points

print("\n" + "="*70)
print("GEODESIC ALIGNMENT VERIFICATION")
print("="*70)
print(f"{'Freq':<10} {'Mean-Min Error':<20} {'Status'}")
print("-"*70)

geodesic_results = {}

for freq in freq_levels:
    errors = []
    geodesics = []
    integrals = []

    # Use first vector field component for testing
    vf = datasets[freq]['vector_field'][:2]  # First fiber direction
    mask = datasets[freq]['mask1']
    metric = results[freq]['metric']

    # FIND VALID POINTS ON THE FIBER
    test_points = find_valid_start_points(mask, num_points=3)

    if not test_points:
        print(f"{freq}x{'':<8} {'No valid points':<20} ✗ FAIL")
        continue

    for pt in test_points:
        # Check initial velocity is non-zero
        vf_np = vf.numpy()
        init_vel = vf_np[:, int(pt[0]), int(pt[1])]

        if np.linalg.norm(init_vel) < 1e-6:
            continue  # Skip points with zero velocity

        err, geo_path, int_path = evaluate_geodesic_alignment(
            metric, vf, mask, pt
        )
        if err < float('inf'):
            errors.append(err)
            geodesics.append(geo_path)
            integrals.append(int_path)

    mean_error = np.mean(errors) if errors else float('inf')
    geodesic_results[freq] = {
        'mean_error': mean_error,
        'geodesics': geodesics,
        'integrals': integrals
    }

    status = "✓ GOOD" if mean_error < 2.0 else "⚠ DRIFT" if mean_error < 5.0 else "✗ FAIL"
    print(f"{freq}x{'':<8} {mean_error:<20.3f} {status}")

print("="*70)
```

```python
# ============================================================
# SIMPLIFIED GEODESIC COMPUTATION (from plot_noise.py)
# ============================================================

def compute_geodesic_2d(metric_lin, vector_lin, mask, start_point, initial_velocity=None,
                        delta_t=0.15, max_iter=2000):
    """Compute geodesic using the geodesic equation γ̈ + Γ(γ̇, γ̇) = 0"""

    # Convert to numpy if needed
    if torch.is_tensor(metric_lin):
        metric_lin = metric_lin.cpu().detach().numpy()
    if torch.is_tensor(vector_lin):
        vector_lin = vector_lin.cpu().detach().numpy()
    if torch.is_tensor(mask):
        mask = mask.cpu().numpy()

    H, W = mask.shape
    metric_mat = tensors.lin2mat(metric_lin)

    # Compute Christoffel symbols
    Gamma1, Gamma2 = riemann.get_christoffel_symbol_2d(metric_mat, mask)

    path_x, path_y = [], []
    gamma = np.array(start_point, dtype=np.float64)

    # Initial velocity from vector field
    if initial_velocity is None:
        x_int, y_int = int(gamma[0]), int(gamma[1])
        if 0 <= x_int < H and 0 <= y_int < W:
            v = np.array([vector_lin[0, x_int, y_int], vector_lin[1, x_int, y_int]])
            v_norm = np.linalg.norm(v)
            gamma_dot = v / v_norm if v_norm > 1e-6 else np.array([1.0, 0.0])
        else:
            gamma_dot = np.array([1.0, 0.0])
    else:
        gamma_dot = np.array(initial_velocity, dtype=np.float64)

    path_x.append(gamma[0])
    path_y.append(gamma[1])

    for i in range(max_iter):
        x, y = gamma
        xi, yi = int(np.clip(x, 0, H-1)), int(np.clip(y, 0, W-1))

        if mask[xi, yi] < 0.5 or x < 1 or x > H-2 or y < 1 or y > W-2:
            break

        # Bilinear interpolation of Christoffel symbols
        x0, y0 = int(x), int(y)
        x1, y1 = min(x0+1, H-1), min(y0+1, W-1)
        fx, fy = x - x0, y - y0

        def interp(field):
            return ((1-fx)*(1-fy)*field[x0, y0] + fx*(1-fy)*field[x1, y0] +
                    (1-fx)*fy*field[x0, y1] + fx*fy*field[x1, y1])

        G1 = np.array([[interp(Gamma1[..., 0, 0]), interp(Gamma1[..., 0, 1])],
                       [interp(Gamma1[..., 1, 0]), interp(Gamma1[..., 1, 1])]])
        G2 = np.array([[interp(Gamma2[..., 0, 0]), interp(Gamma2[..., 0, 1])],
                       [interp(Gamma2[..., 1, 0]), interp(Gamma2[..., 1, 1])]])

        # Geodesic equation
        gamma_ddot = np.zeros(2)
        gamma_ddot[0] = -np.einsum('ij,i,j->', G1, gamma_dot, gamma_dot)
        gamma_ddot[1] = -np.einsum('ij,i,j->', G2, gamma_dot, gamma_dot)

        gamma_dot = gamma_dot + delta_t * gamma_ddot
        gamma = gamma + delta_t * gamma_dot

        path_x.append(gamma[0])
        path_y.append(gamma[1])

    return np.array(path_x), np.array(path_y)


# ============================================================
# UPDATED GEODESIC EVALUATION USING SIMPLIFIED FUNCTION
# ============================================================

def compute_integral_curve(vector_lin, mask, start_point, delta_t=0.1, max_iter=2000):
    """Simple Euler integration along vector field (ground truth)."""
    if torch.is_tensor(vector_lin):
        vector_lin = vector_lin.cpu().numpy()
    if torch.is_tensor(mask):
        mask = mask.cpu().numpy()

    H, W = mask.shape
    path_x, path_y = [], []
    pos = np.array(start_point, dtype=np.float64)

    for _ in range(max_iter):
        xi, yi = int(np.clip(pos[0], 0, H-1)), int(np.clip(pos[1], 0, W-1))
        if mask[xi, yi] < 0.5 or pos[0] < 1 or pos[0] > H-2 or pos[1] < 1 or pos[1] > W-2:
            break

        v = np.array([vector_lin[0, xi, yi], vector_lin[1, xi, yi]])
        if np.linalg.norm(v) < 1e-6:
            break

        path_x.append(pos[0])
        path_y.append(pos[1])
        pos = pos + delta_t * v

    return np.array(path_x), np.array(path_y)


# ============================================================
# RUN VERIFICATION
# ============================================================

print("\n" + "="*70)
print("GEODESIC ALIGNMENT VERIFICATION (Simplified)")
print("="*70)

for freq in freq_levels:
    metric = results[freq]['metric']
    vf = datasets[freq]['vector_field'][:2]
    mask = datasets[freq]['mask1']

    # Convert metric to linear form
    metric_np = metric.numpy()
    metric_lin = tensors.mat2lin(metric_np)

    # Find valid start point
    valid = np.where(mask.numpy() > 0.5)
    mid_idx = len(valid[0]) // 2
    start = np.array([valid[0][mid_idx], valid[1][mid_idx]])

    # Compute geodesic and integral curve
    geo_x, geo_y = compute_geodesic_2d(metric_lin, vf, mask, start, delta_t=0.1, max_iter=500)
    int_x, int_y = compute_integral_curve(vf, mask, start, delta_t=0.1, max_iter=500)

    print(f"\nFreq {freq}x:")
    print(f"  Start: {start}")
    print(f"  Geodesic points: {len(geo_x)}")
    print(f"  Integral points: {len(int_x)}")

    # Compute error if both have points
    if len(geo_x) > 10 and len(int_x) > 10:
        geo_pts = np.column_stack([geo_x, geo_y])
        int_pts = np.column_stack([int_x, int_y])

        min_dists = []
        for gpt in geo_pts:
            dists = np.sqrt(np.sum((int_pts - gpt)**2, axis=1))
            min_dists.append(np.min(dists))

        mean_error = np.mean(min_dists)
        print(f"  Mean-Min Error: {mean_error:.3f}")
    else:
        print(f"  ⚠ Not enough points for comparison")
```

```python
# ============================================================
# DEFINE FREQ LEVELS (if not already defined)
# ============================================================
freq_levels = [1.0, 2.0, 3.0, 4.0]

# ============================================================
# OPTION A: LOCAL CURVATURE PATCH EXPERIMENT
# ============================================================

def generate_local_curvature_patch(high_curv_freq=4.0, img_size=100):
    """
    Single thick fiber with LOCAL high-curvature region.
    Curvature varies only in x ∈ [40, 60], rest is gentle.
    """
    x1 = np.arange(img_size)

    # Frequency that varies locally
    freq = np.ones(img_size) * (np.pi / 4)  # Base gentle frequency

    # HIGH curvature in middle region
    high_region = (x1 >= 40) & (x1 <= 60)
    freq[high_region] = np.pi / 4 * high_curv_freq

    # Integrate to get smooth curve (cumulative angle)
    angle = np.cumsum(freq) / img_size * 2
    curve_y = 20 * np.sin(angle) + 50

    # Derivative for tangent
    dy = np.gradient(curve_y)

    # Create THICK fiber (no narrow regions)
    fiber_width = 10  # Wide fiber

    vf = np.zeros((2, img_size, img_size))
    mask = np.zeros((img_size, img_size))

    for i, x in enumerate(x1):
        y_center = int(np.clip(curve_y[i], 0, img_size-1))
        tangent = np.array([1.0, dy[i]])
        tangent = tangent / (np.linalg.norm(tangent) + 1e-8)

        for d in range(-fiber_width, fiber_width+1):
            y = y_center + d
            if 0 <= y < img_size:
                weight = np.exp(-(d**2) / (fiber_width**2 / 2))
                mask[x, y] = max(mask[x, y], weight)
                vf[0, x, y] += tangent[0] * weight
                vf[1, x, y] += tangent[1] * weight

    # Normalize
    norm = np.sqrt(vf[0]**2 + vf[1]**2)
    norm = np.where(norm > 1e-6, norm, 1)
    vf[0] /= norm
    vf[1] /= norm

    mask_bin = (mask > 0.3).astype(float)

    # Store true curvature for comparison
    # κ = |y''| / (1 + y'^2)^(3/2)
    dy2 = np.gradient(dy)
    true_curvature = np.abs(dy2) / (1 + dy**2)**(1.5)

    return {
        'vector_field': torch.from_numpy(vf).float(),
        'mask': torch.from_numpy(mask_bin).float(),
        'curve_y': curve_y,
        'true_curvature': true_curvature,
        'high_curv_freq': high_curv_freq
    }


def compare_curvatures(data, metric, freq_label):
    """Compare true curvature vs learned geodesic curvature."""

    vf = data['vector_field'][:2] if data['vector_field'].shape[0] == 4 else data['vector_field']
    mask = data['mask'] if 'mask' in data else data['mask1']

    metric_np = metric.numpy()
    metric_lin = tensors.mat2lin(metric_np)

    # Compute geodesic curvature from learned metric
    kappa_geo = compute_geodesic_curvature(metric_lin, vf, mask)

    # Get true curvature along the fiber centerline
    true_curv = data.get('true_curvature', None)

    return kappa_geo, true_curv


# ============================================================
# RUN OPTION A: LOCAL CURVATURE PATCH
# ============================================================

print("\n" + "="*70)
print("OPTION A: LOCAL CURVATURE PATCH EXPERIMENT")
print("="*70)

# Generate patches with different local curvature intensities
local_curv_levels = [1.0, 2.0, 4.0, 8.0]
patch_datasets = {}

for lc in local_curv_levels:
    patch_datasets[lc] = generate_local_curvature_patch(high_curv_freq=lc)
    print(f"Generated patch with local curvature {lc}x in region [40,60]")

# Visualize
fig, axes = plt.subplots(2, len(local_curv_levels), figsize=(16, 6))
for idx, lc in enumerate(local_curv_levels):
    d = patch_datasets[lc]

    # Top: Fiber with curvature region highlighted
    ax = axes[0, idx]
    ax.imshow(d['mask'].numpy().T, origin='lower', cmap='gray', alpha=0.5)
    ax.plot(range(100), d['curve_y'], 'b-', linewidth=2)
    ax.axvspan(40, 60, alpha=0.3, color='red', label='High curv region')
    ax.set_title(f'Local {lc}x curvature')
    ax.set_xlim(0, 100)
    ax.set_ylim(0, 100)

    # Bottom: True curvature profile
    ax = axes[1, idx]
    ax.plot(range(100), d['true_curvature'], 'r-', linewidth=2)
    ax.axvspan(40, 60, alpha=0.2, color='red')
    ax.set_xlabel('x')
    ax.set_ylabel('κ_true')
    ax.set_title('True curvature')

plt.suptitle('Option A: Local Curvature Patch (Same thick fiber, varying local curvature)',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('../plots/local_curvature_patch.png', dpi=150)
plt.show()


# ============================================================
# CURVATURE COMPARISON FOR ORIGINAL BRAIDS
# ============================================================

print("\n" + "="*70)
print("CURVATURE ANALYSIS: κ_geo vs κ_true")
print("="*70)

fig, axes = plt.subplots(2, len(freq_levels), figsize=(16, 6))

for idx, freq in enumerate(freq_levels):
    d = datasets[freq]
    metric = results[freq]['metric']

    vf = d['vector_field'][:2]
    mask = d['mask1']

    # Compute geodesic curvature
    metric_np = metric.numpy()
    metric_lin = tensors.mat2lin(metric_np)
    kappa_geo = compute_geodesic_curvature(metric_lin, vf, mask)

    # True curvature for parametric curve: κ = freq² * amplitude
    base_freq = np.pi / 4 * freq
    x1 = np.arange(100)
    curve1_x2 = 20 * np.cos(base_freq * (x1 - 60)) + 50
    dy = -20 * base_freq * np.sin(base_freq * (x1 - 60))
    dy2 = -20 * base_freq**2 * np.cos(base_freq * (x1 - 60))
    kappa_true_curve = np.abs(dy2) / (1 + dy**2)**(1.5)

    # Top row: κ_geo heatmap
    ax = axes[0, idx]
    kappa_masked = kappa_geo * mask.numpy()
    im = ax.imshow(kappa_masked.T, origin='lower', cmap='hot', vmin=0, vmax=0.5)
    ax.set_title(f'Freq {freq}x\nκ_geo (learned)')
    ax.axis('off')

    # Bottom row: κ_true vs κ_geo along centerline
    ax = axes[1, idx]
    ax.plot(x1, kappa_true_curve, 'b-', label='κ_true', linewidth=2)

    # Sample κ_geo along the curve
    kappa_geo_samples = []
    for i, x in enumerate(x1):
        y = int(np.clip(curve1_x2[i], 0, 99))
        kappa_geo_samples.append(kappa_geo[x, y])
    ax.plot(x1, kappa_geo_samples, 'r--', label='κ_geo', linewidth=2, alpha=0.7)

    ax.set_xlabel('x')
    ax.set_ylabel('Curvature')
    ax.legend(fontsize=8)
    ax.set_ylim(0, max(kappa_true_curve) * 1.5)

plt.colorbar(im, ax=axes[0, :], shrink=0.6, label='κ_geo')
plt.suptitle('Curvature Comparison: Does the Learned Metric Capture True Curvature?',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('../plots/curvature_comparison.png', dpi=150, bbox_inches='tight')
plt.show()

# Summary statistics
print("\nCurvature Accuracy Summary:")
print("-"*50)
for freq in freq_levels:
    d = datasets[freq]
    metric = results[freq]['metric']

    vf = d['vector_field'][:2]
    mask = d['mask1']
    metric_np = metric.numpy()
    metric_lin = tensors.mat2lin(metric_np)

    kappa_geo = compute_geodesic_curvature(metric_lin, vf, mask)

    # Compute true curvature
    base_freq = np.pi / 4 * freq
    x1 = np.arange(100)
    curve1_x2 = 20 * np.cos(base_freq * (x1 - 60)) + 50
    dy = -20 * base_freq * np.sin(base_freq * (x1 - 60))
    dy2 = -20 * base_freq**2 * np.cos(base_freq * (x1 - 60))
    kappa_true_curve = np.abs(dy2) / (1 + dy**2)**(1.5)

    # Sample and compare
    kappa_geo_samples = []
    kappa_true_samples = []
    for i, x in enumerate(x1):
        y = int(np.clip(curve1_x2[i], 0, 99))
        if mask.numpy()[x, y] > 0.5:
            kappa_geo_samples.append(kappa_geo[x, y])
            kappa_true_samples.append(kappa_true_curve[i])

    if kappa_geo_samples:
        ratio = np.mean(kappa_geo_samples) / (np.mean(kappa_true_samples) + 1e-8)
        correlation = np.corrcoef(kappa_geo_samples, kappa_true_samples)[0, 1]

        status = "underestimates" if ratio < 0.8 else "overestimates" if ratio > 1.2 else "matches"
        print(f"Freq {freq}x: κ_geo/κ_true = {ratio:.2f} ({status}), correlation = {correlation:.3f}")
```

```python
# ============================================================
# OPTION B: FIBER THICKNESS ROBUSTNESS TEST
# ============================================================
# Question: How much spatial support does the model need?
# Method: Train on same curvature (1.0x), vary mask thickness

from scipy.ndimage import binary_dilation, binary_erosion

def generate_braid_with_thickness(thickness_factor=1.0, freq=1.0, img_size=100):
    """Generate braid with controlled fiber thickness."""

    base_freq = np.pi / 4 * freq
    x1 = np.arange(img_size)

    curve1_x2 = 20 * np.cos(base_freq * (x1 - 60)) + 50
    curve2_x2 = 20 * np.sin(base_freq * x1) + 50

    curve1_dx2 = -20 * base_freq * np.sin(base_freq * (x1 - 60))
    curve2_dx2 = 20 * base_freq * np.cos(base_freq * x1)

    # BASE fiber width, then dilate/erode
    base_width = 4

    vf = np.zeros((4, img_size, img_size))
    mask1 = np.zeros((img_size, img_size))
    mask2 = np.zeros((img_size, img_size))

    # Draw thin base fibers first
    for i, x in enumerate(x1):
        y1 = int(np.clip(curve1_x2[i], 0, img_size-1))
        y2 = int(np.clip(curve2_x2[i], 0, img_size-1))

        t1 = np.array([1.0, curve1_dx2[i]])
        t1 = t1 / (np.linalg.norm(t1) + 1e-8)
        t2 = np.array([1.0, curve2_dx2[i]])
        t2 = t2 / (np.linalg.norm(t2) + 1e-8)

        for dy in range(-base_width, base_width+1):
            if 0 <= y1+dy < img_size:
                weight = np.exp(-(dy**2) / (base_width**2))
                mask1[x, y1+dy] = max(mask1[x, y1+dy], weight)
                vf[0, x, y1+dy] += t1[0] * weight
                vf[1, x, y1+dy] += t1[1] * weight
            if 0 <= y2+dy < img_size:
                weight = np.exp(-(dy**2) / (base_width**2))
                mask2[x, y2+dy] = max(mask2[x, y2+dy], weight)
                vf[2, x, y2+dy] += t2[0] * weight
                vf[3, x, y2+dy] += t2[1] * weight

    # Normalize VF
    for c in [0, 2]:
        norm = np.sqrt(vf[c]**2 + vf[c+1]**2)
        norm = np.where(norm > 1e-6, norm, 1)
        vf[c] /= norm
        vf[c+1] /= norm

    # Apply thickness factor via dilation/erosion
    mask1_bin = (mask1 > 0.3).astype(float)
    mask2_bin = (mask2 > 0.3).astype(float)

    if thickness_factor > 1.0:
        iterations = int((thickness_factor - 1) * 3)
        mask1_bin = binary_dilation(mask1_bin, iterations=iterations).astype(float)
        mask2_bin = binary_dilation(mask2_bin, iterations=iterations).astype(float)
    elif thickness_factor < 1.0:
        iterations = int((1 - thickness_factor) * 3)
        mask1_bin = binary_erosion(mask1_bin, iterations=iterations).astype(float)
        mask2_bin = binary_erosion(mask2_bin, iterations=iterations).astype(float)

    return {
        'vector_field': torch.from_numpy(vf).float(),
        'mask1': torch.from_numpy(mask1_bin).float(),
        'mask2': torch.from_numpy(mask2_bin).float(),
        'combined_mask': torch.from_numpy(np.clip(mask1_bin + mask2_bin, 0, 1)).float(),
        'thickness': thickness_factor
    }

# Test different thicknesses
thickness_levels = [0.5, 0.75, 1.0, 1.5, 2.0]

print("="*60)
print("FIBER THICKNESS ROBUSTNESS TEST")
print("="*60)

thickness_results = {}

for thickness in thickness_levels:
    data = generate_braid_with_thickness(thickness_factor=thickness)
    coverage = data['mask1'].sum().item()
    print(f"Thickness {thickness}x: {coverage:.0f} pixels")
    thickness_results[thickness] = data

# Visualize
fig, axes = plt.subplots(1, len(thickness_levels), figsize=(15, 3))
for idx, t in enumerate(thickness_levels):
    ax = axes[idx]
    ax.imshow(thickness_results[t]['combined_mask'].numpy().T, origin='lower', cmap='gray')
    ax.set_title(f'Thickness {t}x\n{thickness_results[t]["mask1"].sum():.0f} px')
    ax.axis('off')
plt.suptitle('Fiber Thickness Variations (Fixed Curvature)', fontsize=12)
plt.tight_layout()
plt.savefig('../plots/thickness_variations.png', dpi=150)
plt.show()
```

```python
# ============================================================
# OPTION C: GEODESIC PARALLELISM / DIVERGENCE TEST
# ============================================================
# Question: Do geodesics from nearby seeds stay parallel or diverge?

def test_geodesic_parallelism(metric_lin, vector_lin, mask, num_seeds=5):
    """
    Start multiple geodesics along the same fiber and measure divergence.
    Returns average pairwise Hausdorff distance.
    """
    # Find valid starting points along the fiber
    valid = np.where(mask > 0.5)
    if len(valid[0]) < num_seeds * 10:
        return float('inf'), []

    # Sample seed points evenly along the fiber
    indices = np.linspace(len(valid[0])//4, 3*len(valid[0])//4, num_seeds).astype(int)
    seeds = [np.array([valid[0][i], valid[1][i]]) for i in indices]

    # Compute geodesics from each seed
    geodesics = []
    for seed in seeds:
        gx, gy = compute_geodesic_2d(metric_lin, vector_lin, mask, seed,
                                      delta_t=0.05, max_iter=500)
        if len(gx) > 20:
            geodesics.append((gx, gy))

    if len(geodesics) < 2:
        return float('inf'), geodesics

    # Compute pairwise Hausdorff distances
    hausdorff_dists = []
    for i in range(len(geodesics)):
        for j in range(i+1, len(geodesics)):
            pts1 = np.column_stack(geodesics[i])
            pts2 = np.column_stack(geodesics[j])

            d1 = np.max([np.min(np.sqrt(np.sum((pts2 - p)**2, axis=1))) for p in pts1])
            d2 = np.max([np.min(np.sqrt(np.sum((pts1 - p)**2, axis=1))) for p in pts2])
            hausdorff_dists.append(max(d1, d2))

    return np.mean(hausdorff_dists), geodesics

# Run on each frequency level
print("\n" + "="*60)
print("GEODESIC PARALLELISM TEST")
print("="*60)
print("(Lower divergence = more stable metric)")
print("-"*60)

parallelism_results = {}
freq_levels = [1.0, 2.0, 3.0, 4.0]

for freq in freq_levels:
    metric = results[freq]['metric'].numpy()
    metric_lin = tensors.mat2lin(metric)
    vf = datasets[freq]['vector_field'][:2].numpy()
    mask = datasets[freq]['mask1'].numpy()

    divergence, geos = test_geodesic_parallelism(metric_lin, vf, mask, num_seeds=5)
    parallelism_results[freq] = {'divergence': divergence, 'geodesics': geos}

    status = "✓ Parallel" if divergence < 3 else "⚠ Drifting" if divergence < 10 else "✗ Diverging"
    print(f"Freq {freq}x: Divergence = {divergence:.2f} {status}")

# Visualize geodesic fans
fig, axes = plt.subplots(1, len(freq_levels), figsize=(16, 4))
colors = plt.cm.viridis(np.linspace(0, 1, 5))

for idx, freq in enumerate(freq_levels):
    ax = axes[idx]
    mask = datasets[freq]['combined_mask'].numpy()
    ax.imshow(mask.T, origin='lower', cmap='gray', alpha=0.3)

    for i, (gx, gy) in enumerate(parallelism_results[freq]['geodesics']):
        ax.plot(gx, gy, color=colors[i], linewidth=2, alpha=0.8)

    div = parallelism_results[freq]['divergence']
    ax.set_title(f'Freq {freq}x\nDivergence: {div:.1f}')
    ax.axis('off')

plt.suptitle('Geodesic Parallelism Test: Do Nearby Seeds Stay Together?',
             fontsize=12, fontweight='bold')
plt.tight_layout()
plt.savefig('../plots/geodesic_parallelism.png', dpi=150, bbox_inches='tight')
plt.show()

# Summary trend
fig, ax = plt.subplots(figsize=(8, 5))
freqs = list(parallelism_results.keys())
divs = [parallelism_results[f]['divergence'] for f in freqs]
divs = [d if d < 100 else np.nan for d in divs]  # Cap inf values

ax.plot(freqs, divs, 'o-', linewidth=2, markersize=10, color='#E63946')
ax.set_xlabel('Frequency Multiplier')
ax.set_ylabel('Geodesic Divergence (Hausdorff)')
ax.set_title('Geodesic Stability vs Curvature')
ax.grid(True, alpha=0.3)
plt.savefig('../plots/divergence_trend.png', dpi=150)
plt.show()
```

```python
class ImageDataset(Dataset):
    def __init__(self, data_dir):
        self.data_dir = data_dir

    def __len__(self):
        return 1

    def __getitem__(self, idx):
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
        vector_field1_path = f'{self.data_dir}/sin/sin_vector_field.nhdr'
        mask1_path = f'{self.data_dir}/sin/sin_filt_mask.nhdr'
        vector_field1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field1_path))).permute(2,0,1).to(device).float()*1000.0
        mask1 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask1_path))).permute(1,0)
        boundary_mask1 = torch.where(torch.from_numpy(filters.laplace(mask1))>0,1,0)
        mask1 = (mask1-boundary_mask1).float().to(device)

        vector_field2_path = f'{self.data_dir}/cos/cos_vector_field.nhdr'
        mask2_path = f'{self.data_dir}/cos/cos_filt_mask.nhdr'
        vector_field2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(vector_field2_path))).permute(2,0,1).to(device).float()*1000.0
        mask2 = torch.from_numpy(sitk.GetArrayFromImage(sitk.ReadImage(mask2_path))).permute(1,0)
        boundary_mask2 = torch.where(torch.from_numpy(filters.laplace(mask2))>0,1,0)
        mask2 = (mask2-boundary_mask2).float().to(device)

        sample = {  'vector_field'  : torch.cat((vector_field1,vector_field2),0),
                    'mask1'          : mask1.unsqueeze(0),
                    'mask2'          : mask2.unsqueeze(0)}
        return sample

def matrix_exp_2d(A):
    """
    Construct positive definite matrix from symmetric matrix field A
    Args:
        A, torch.Tensor
    Returns:
        psd, torch.Tensor
    """
    I = torch.zeros_like(A, device='cuda')
    I[...,0,0] = 1
    I[...,1,1] = 1

    s = ((A[...,0,0]+A[...,1,1])/2.).unsqueeze(-1).unsqueeze(-1)
    q = torch.sqrt(-torch.det(A-torch.mul(s, I))).unsqueeze(-1).unsqueeze(-1)

    psd = torch.exp(s)*(torch.mul((torch.cosh(q)-s*torch.sinh(q)/q),I)+torch.sinh(q)/q*A)
    return psd

def pde(u, vector_lin, mask, differential_accuracy=2):
    s = tensors.lin2mat(u)
    metric_mat = matrix_exp_2d(s)
    nabla_vv = riemann.covariant_derivative_2d(vector_lin, metric_mat, mask, differential_accuracy=differential_accuracy)

    return nabla_vv
```

### [Matrix exponential for 2D](https://en.wikipedia.org/wiki/Matrix_exponential)
${\displaystyle e^{tA}=e^{st}\left(\left(\cosh(qt)-s{\frac {\sinh(qt)}{q}}\right)~I~+{\frac {\sinh(qt)}{q}}A\right)~.}$ where $t=1$, $s=tr({A}/2), q=-\sqrt{-\det(A-sI)}$

# Training

```python
resume = False
save_model = True
print(f'resume:{resume}, save_model:{save_model}')
```

```python
epoch_loss_list = []
epoch_num = 1001
start_epoch_num = 1001
learning_rate = 1e-2

model = DenseED(in_channels=4, out_channels=3,
                imsize=100,
                blocks=[6, 8, 6],
                growth_rate=16,
                init_features=48,
                drop_rate=0,
                out_activation=None,
                upsample='nearest')
model.train()
if mode=='gpu':
    model.cuda()
criterion = torch.nn.MSELoss()
optimizer_id = torch.optim.Adadelta(model.parameters(), lr=learning_rate)

dataset_id = ImageDataset(input_dir)
dataloader_id = DataLoader(dataset_id, batch_size=1, shuffle=True, num_workers=0)

if resume:
    checkpoint = torch.load(f'{output_dir}/epoch_{start_epoch_num-1}_checkpoint.pth.tar')
    model.load_state_dict(checkpoint['model_state_dict'])

    with open(f'{output_dir}/loss.txt', 'a') as f:
        f.write(f'From {start_epoch_num} to {epoch_num+start_epoch_num}\n')
        f.write(f'MSE; Adadelta, lr={learning_rate};\n')
else:
    start_epoch_num = 0

    with open(f'{output_dir}/loss.txt', 'w+') as f:
        f.write(f'From {start_epoch_num} to {epoch_num+start_epoch_num}\n')
        f.write(f'MSE; Adadelta: lr={learning_rate};\n')

print(f'Starting from iteration {start_epoch_num} to iteration {epoch_num+start_epoch_num}')
```

```python
for epoch in tqdm(range(start_epoch_num, start_epoch_num+epoch_num)):
    epoch_loss_id = 0
    epoch_loss_bd = 0

    for i, batched_id_sample in enumerate(dataloader_id):
        '''innerdomain backpropagate'''
        input_id = batched_id_sample['vector_field'].to(device)
        input_id.requires_grad = True
        # u_pred_id: [batch_size, *data_shape, feature_num] = [1, 3, ...]
        u_pred_id = model(input_id)
        # f_pred_id: [batch_size, feature_num, *data_shape] = [1, 2, ...]
        mask1 = batched_id_sample['mask1'].squeeze()
        mask2 = batched_id_sample['mask2'].squeeze()
        pde_loss1 = pde(u_pred_id.squeeze(), input_id[0,:2].squeeze(), mask1, differential_accuracy=2)
        pde_loss2 = pde(u_pred_id.squeeze(), input_id[0,2:].squeeze(), mask2, differential_accuracy=2)
        f_pred_1 = torch.einsum('...ij,...ij->...ij', pde_loss1, mask1.unsqueeze(0).expand(2,-1,-1))
        f_pred_2 = torch.einsum('...ij,...ij->...ij', pde_loss2, mask2.unsqueeze(0).expand(2,-1,-1))
        f_pred_id = torch.cat((f_pred_1,f_pred_2),0)
        # f_true_id: [batch_size, feature_num, *data_shape] = [1, 2, ...]
        f_true_id = torch.zeros_like(f_pred_id)

        optimizer_id.zero_grad()
        loss_id = criterion(f_pred_id, f_true_id)
        loss_id.backward()
        epoch_loss_id += loss_id.item()
        optimizer_id.step()

    with open(f'{output_dir}/loss.txt', 'a') as f:
        f.write(f'{epoch_loss_id}\n')

    print(f'epoch {epoch} innerdomain loss: {epoch_loss_id}, norm: {torch.norm(f_pred_id,2)**2}')
    epoch_loss_list.append(epoch_loss_id)
    if epoch%10==0:
        if save_model:
            torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_id_state_dict': optimizer_id.state_dict(),
            'loss_id': epoch_loss_id
            }, f'{output_dir}/model.pth.tar')
```

```python
plt.figure(figsize=(7,5))
plt.xlabel('Iteration')
plt.ylabel('MSE Loss')
plt.plot(epoch_loss_list)
plt.savefig(f'{output_dir}/adadelta_loss_{learning_rate}.png')
```
